## Statistics Dərs 4 | 2-ci hissə

```python
from PIL import Image

# Fayl yolunun əsas hissəsi
base_path = r'C:\Users\dell\Desktop\Statistics\ders 4\ders_4_'

# Şəkilləri saxlayacağımız siyahı
image_list = []

# 1-dən 21-ə qədər olan şəkilləri dövr ilə açırıq
for i in range(21, 41):  # 1-dən 5-ə qədər
    full_path = f'{base_path}{i}.png'  # Fayl yolunu dinamik olaraq yaradırıq
    img = Image.open(full_path)
    image_list.append(img)  # Şəkili siyahıya əlavə edirik

# Nəticə: images siyahısında bütün yüklənmiş şəkillər olacaq

import matplotlib.pyplot as plt

# Şəkilləri göstərin
for image in image_list:
    plt.figure(figsize=(25, 5))  # Şəkilin ölçüsü
    plt.title('Statistics Dərs 4 | 2-ci hissə', size=18)
    plt.imshow(image)
    plt.axis('off')  # Oxları gizlətmək üçün
    plt.show()
```
