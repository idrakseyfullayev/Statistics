## **"İki filialda hər şey eynidirsə, ortalama da eynidirsə, niyə two sample test aparım?"**




### ✅ Cavab:

**Çünki "hər şey eynidir" və "ortalama da eynidir" deyə düşünmək subyektiv və ya təsadüfi müşahidə ola bilər. Statistika isə subyektivliyə yox, sübuta əsaslanır.**

---

### 📌 Two sample test (məsələn, t-test) nə üçündür?

İki qrupun (məsələn, filial A və filial B-nin) **real fərqinin olub-olmadığını** yoxlamaq üçündür.
Sən deyə bilərsən:

* "Ortalama satış bənzərdi"
  Amma **bu bənzərlik təsadüfdürmü, yoxsa doğrudan da statistik olaraq fərq yoxdur?** — bunu test edir.

---

### 🎯 Niyə test aparılır?

| Səbəb                                          | İzah                                                                                                                     |
| ---------------------------------------------- | ------------------------------------------------------------------------------------------------------------------------ |
| **1. Təsadüfi fərqləri ayırmaq üçün**          | Ola bilər ki, bu ay filial A 10₼ çox qazandı. Bu **real fərq**dirmi, yoxsa **şans əsəridirmi**?                          |
| **2. Elmi əsaslı qərar üçün**                  | Gələcəkdə hansı filialın daha çox investisiya alması, hansı strategiyanın uğurlu olduğu statistik sübuta əsaslanmalıdır. |
| **3. Tək ortalama deyil, yayılma da vacibdir** | Ortalamalar eyni ola bilər, amma biri stabildir (dəyərlər yaxın), digəri dəyişkəndir.                                    |
| **4. Gələcək fərqləri proqnozlaşdırmaq üçün**  | Əgər test fərq olmadığını göstərirsə, deməli, gələcəkdə də fərq olma ehtimalı azdır.                                     |

---

### 🧪 Məsələn:

| Filial | Ortalama Satış (₼) | Std sapma | n (nümunə sayı) |
| ------ | ------------------ | --------- | --------------- |
| A      | 100                | 12        | 30              |
| B      | 101                | 15        | 30              |

Burda **ortalama fərq cəmi 1₼**-dir. Gözlə baxanda "eyni" görünür.
Amma two sample t-test deyə bilər:

* "Bu fərq statistik olaraq **əhəmiyyətsizdir**" → Doğrudan da eynidir.
* Yaxud da "Bu fərq **əhəmiyyətlidir**" → Deməli, bu kiçik fərq təsadüfi deyil.

---

### 📌 Nəticə:

> “Eyni görünür” ≠ “Statistik olaraq eynidir”
> Bunu **test etməsən**, yalnız **intuisiya ilə** qərar verirsən.




## İki filialın ortalaması eyni p-value < 0.05



### ❓ Sual:

- İki filialın ortalaması **eyni çıxıb**
- Amma **p-value < 0.05** çıxıb.
- **Necə olur bu? Ortalamalar eynidirsə, niyə fərq əhəmiyyətli çıxsın?**

---

### ✅ Cavab:

**Ortalamaların "eyni çıxması" tam dəqiq eyni rəqəm demək deyil — çox yaxın olması da belə başa düşülür. Amma statistik testlər fərqin təsadüfi olub-olmadığını yoxlayır.**

Yəni:

> 🔍 **Orta fərq çox kiçik olsa belə**,
> əgər **nümunə sayı böyükdürsə** və ya **standart sapma çox azdırsa**,
> onda bu **kiçik fərq belə əhəmiyyətli** (significant) çıxır.

---

### 📊 Texniki olaraq bu baş verə bilər:

#### Məsələn:

| Filial | Ortalama | Std | n    |
| ------ | -------- | --- | ---- |
| A      | 100.00   | 0.5 | 1000 |
| B      | 100.10   | 0.5 | 1000 |

Bu fərq cəmi **0.1₼** olsa da, **p-value < 0.05** çıxacaq. Çünki:

* Nümunə sayı böyükdür
* Std sapma azdır
* Belə olan halda, **hətta balaca fərq belə statistik olaraq əhəmiyyətli** görünür.

---

### 📌 Qısa qayda:

| Hal                                           | Nəticə                                                                                |
| --------------------------------------------- | ------------------------------------------------------------------------------------- |
| **Ortalama fərqi az, amma p-value < 0.05**    | Kiçik fərq **statistik olaraq əhəmiyyətlidir**, çünki **n böyük və ya yayılma azdır** |
| **Ortalama fərqi böyük, amma p-value > 0.05** | Böyük fərq **statistik deyil**, çünki **n az və ya yayılma çoxdur**                   |

---

### 🧠 Nəticə:

> **Orta dəyərlərə baxmaq kifayət deyil.**
> **p-value bizə deyir: bu fərq təsadüfidirmi, yoxsa real?**




## Hipotez Testlərinin Əsas Fəlsəfəsi



### 📌 1. **Ortalama eyni çıxıb, amma p-value < 0.05**

> ❓ Bu nə deməkdir?

### ✅ Açıqlama:

* Ortalamalar **təxminən eynidir**, fərq bəlkə də **0.1₼**, **0.01₼** kimi çox **kiçik bir şeydir**.
* Amma **statistik test** deyir ki:

> “Bu kiçik fərq **təsadüfən** yaranmayıb. Əgər H0 (eyni olduqları fərziyyəsi) doğru olsaydı, bu nəticə **çox nadir** baş verərdi.”

### 🎯 Nəticə:

> **Deməli, əslində bu iki filial fərqli işləyib. Sadəcə ortalamaları təsadüfən eyni çıxıb.**

### 🧠 Analogi:

- Bu sanki **iki tələbənin** imtahandan **eyni bal alması** kimidir.
- Amma biri **çox çətin sualları** düz yazıb, digəri **asan suallar**. Ortalamaları eyni çıxsa da, **daxili keyfiyyətləri fərqlidir.**

---

### 📌 2. **Ortalama fərqlidir, amma p-value > 0.05**

> ❓ Bu nə deməkdir?

### ✅ Açıqlama:

* Bu halda **sən görünən fərq** görürsən (məsələn, A filialı 105₼, B filialı 100₼).
* Amma **test deyir**: “Bu fərq **statistik olaraq əhəmiyyətli deyil**. Yəni bu fərq **təsadüfən** də yaranmış ola bilər.”
* Ola bilər ki:

  * **Nümunə sayı azdır** (n kiçikdirsə, test zəif olur)
  * **Dəyişkənlik çoxdur** (standart sapma böyükdür)

### 🎯 Nəticə:

> **Deməli, bu görünən fərq real deyil, sadəcə “şans əsəri” ola bilər. Əslində performanslar bənzərdir.**

---

### 📊 İndi bu ikisini cədvəl ilə ümumiləşdirək:

| Ortalama Fərqi   | P-value      | Nəticə                                                                                 |
| ---------------- | ------------ | -------------------------------------------------------------------------------------- |
| Eyni / çox yaxın | **p < 0.05** | **Təsadüfən eyni çıxıb. Əslində fərq var.**                                            |
| Fərqli           | **p > 0.05** | **Fərq təsadüfidir. Əslində fərq yoxdur.**                                             |
| Eyni             | **p > 0.05** | Eyni çıxıb, p-value da böyük → **Doğrudan da fərq yoxdur.**                            |
| Fərqli           | **p < 0.05** | Ortalamalar fərqlidir və **statistik olaraq da bu əhəmiyyətlidir** → **Əsl fərq var.** |

---

### 🧪 Statistika sənə bunu deməyə kömək edir:

> “Ortalamaya baxıb özün qərar vermə! Mən sənə deyim ki, bu fərq **real**dır, yoxsa **şans əsəridir**.”

* **Nə üçün:** 3+ qrup arasında fərq varmı?
* ✅ **Nümunə:** Bakı, Gəncə, Sumqayıt filiallarının ortalama gündəlik gəliri eynidirmi?

```python
from scipy.stats import f_oneway

# Gəlir dataları
baku = [100, 110, 95, 105]
ganca = [90, 85, 88, 92]
sumqayit = [102, 98, 100, 101]

f_stat, p_val = f_oneway(baku, ganca, sumqayit)
print(f"F-stat: {f_stat}, p-value: {p_val}")
```

### 2. **Anova: Two-Factor With Replication**

* **Nədir:** İki faktorun təsiri və qarşılıqlı təsirini ölçür.
* ✅ **Nümunə:** 3 fərqli dərman, 4 müxtəlif gün ərzində qan təzyiqinə necə təsir edir?

```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Data çərçivəsi (Dərman, Gün, Təzyiq)
df = pd.read_csv('your_data.csv')  # sütunlar: Drug, Day, BP

model = ols('BP ~ C(Drug) + C(Day) + C(Drug):C(Day)', data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
print(anova_table)
```

### 16. **t-Test: Paired Two Sample for Means**

* **Nədir:** Eyni qrupun əvvəl və sonra ölçülərinin fərqi.
* ✅ **Nümunə:** Təlimdən əvvəl və sonra test balları.

```python
from scipy.stats import ttest_rel

before = [80, 85, 78, 90]
after = [85, 88, 82, 95]

t_stat, p = ttest_rel(before, after)
print(f"T-stat: {t_stat}, p-value: {p}")
```

### 17. **t-Test: Two Sample Assuming Equal Variances**

* **Nədir:** İki qrupun ortalaması eynidirmi? Varianslar bərabərdir.
* ✅ **Nümunə:** Kişi və qadın maaşı fərqlidirmi?

```python
from scipy.stats import ttest_ind

male = [2000, 2200, 2100]
female = [1900, 2000, 1950]

t_stat, p = ttest_ind(male, female, equal_var=True)
print(f"T-stat: {t_stat}, p-value: {p}")
```

---

### 18. **t-Test: Two Sample Assuming Unequal Variances**

* **Nədir:** Varianslar fərqli olsa da ortalamaları müqayisə edir.
* ✅ **Nümunə:** İki məktəbdə ortalama bal fərqlidirmi?

```python
from scipy.stats import ttest_ind

school1 = [80, 85, 83]
school2 = [75, 70, 78]

t_stat, p = ttest_ind(school1, school2, equal_var=False)
print(f"T-stat: {t_stat}, p-value: {p}")
```
