## $y = b_0 + b_1 \cdot x$



$$
y = b_0 + b_1 \cdot x
$$

Bu **xətti regresiya** və ya **xətti funksiyanın ümumi formasıdır**. Buradakı hər simvolun mənası belədir:

---

### 🔹 1. $y$ — Asılı dəyişən (dependent variable)

* **Nəticə, proqnoz edilən dəyər**.
* Bizim tapmaq istədiyimiz, izah etmək istədiyimiz göstəricidir.
* Məsələn: "Evin qiyməti" $y$ ola bilər.

---

### 🔹 2. $x$ — Müstəqil dəyişən (independent variable)

* **Təsir edən amil, giriş dəyişəni**.
* Biz $y$-nin dəyişməsini bu dəyişənlə izah edirik.
* Məsələn: "Evin sahəsi (kvadrat metr)" $x$ ola bilər.

---

### 🔹 3. $b_0$ — İntersept (sabit termin, intercept)

* **x = 0 olduqda y-in dəyəri**.
* Qrafikdə düz xəttin **y oxunu kəsdiyi nöqtədir**.
* Məsələn: evin sahəsi 0 olsa belə (teorik olaraq), evin müəyyən başlanğıc qiyməti ola bilər. Bu başlanğıc dəyəri $b_0$-dır.

---

### 🔹 4. $b_1$ — Əmsal (slope, regresiya əmsalı)

* **Xəttin meyli (slope)**.
* $x$-dəki **hər vahid artım üçün $y$-də orta hesabla nə qədər dəyişiklik olduğunu göstərir**.
* Riyazi olaraq:

  $$
  b_1 = \frac{\Delta y}{\Delta x}
  $$

  yəni $x$ 1 vahid artırsa, $y$ təxminən $b_1$ qədər dəyişir.
* Məsələn: evin sahəsi hər 1 kv.m artdıqca qiymət 500 manat artırsa, $b_1 = 500$.

---

### 🔹 Sadə Misal:

$$
y = 20000 + 500 \cdot x
$$

* $b_0 = 20000$ → evin minimal başlanğıc qiyməti (sahəsi 0 olsa belə).
* $b_1 = 500$ → hər kvadrat metr üçün qiymət 500 manat artır.
* $x = 50$ kv.m ev üçün:

  $$
  y = 20000 + 500 \cdot 50 = 20000 + 25000 = 45000
  $$
