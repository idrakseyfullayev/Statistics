## Regression Modeldə Ən kiçik kvadratlar OLS həlli



### 1️⃣ Modelin başlanğıc forması:

$$
y = \beta_0 + \beta_1 X + \varepsilon
$$

---

### 2️⃣ Vektor-matris formasında yazılışı:

$$
\begin{bmatrix}
y_1 \\
y_2 \\
\vdots \\
y_n
\end{bmatrix}
=
\beta_0
\begin{bmatrix}
1 \\
1 \\
\vdots \\
1
\end{bmatrix}
+
\beta_1
\begin{bmatrix}
x_1 \\
x_2 \\
\vdots \\
x_n
\end{bmatrix}
+
\begin{bmatrix}
\varepsilon_1 \\
\varepsilon_2 \\
\vdots \\
\varepsilon_n
\end{bmatrix}
$$

---

### 3️⃣ Ümumi **matris forması**:

$$
y =
\begin{bmatrix}
1 & x_1 \\
1 & x_2 \\
\vdots & \vdots \\
1 & x_n
\end{bmatrix}
\begin{bmatrix}
\beta_0 \\
\beta_1
\end{bmatrix}
+
\begin{bmatrix}
\varepsilon_1 \\
\varepsilon_2 \\
\vdots \\
\varepsilon_n
\end{bmatrix}
$$

yəni:

$$
y = X\beta + \varepsilon
$$

---

### 4️⃣ Ən kiçik kvadratlar (OLS) həlli:

$$
\hat{\beta} = (X^T X)^{-1} X^T y
$$




## **Xətti reqressiyanın (Linear Regression) (Ordinary Least Squares - OLS)** ilə tapılan həlli



### 1️⃣ Xətti modelin qurulması

Bizim xətti reqressiya modelimiz belədir:

$$
y = X \beta + \varepsilon
$$

Burada:

* $y$ → müşahidə olunan nəticələrin vektoru (n × 1 ölçülü),
* $X$ → atribut (features) matrisi (n × p ölçülü),
* $\beta$ → modelin parametrləri (p × 1 ölçülü),
* $\varepsilon$ → səhv (error, residual) vektoru.

---

### 2️⃣ Məqsəd funksiyası (OLS kriteriyası)

Ən kiçik kvadratlar üsulunda məqsəd odur ki, səhvlərin kvadratlarının cəmini minimallaşdıraq:

$$
S(\beta) = \| y - X \beta \|^2
$$

---

### 3️⃣ Kvadrat ifadəni açırıq

Normanı açaq:

$$
S(\beta) = (y - X\beta)^T (y - X\beta)
$$

$$
= y^T y - y^T X \beta - \beta^T X^T y + \beta^T X^T X \beta
$$

Qeyd: Orta iki ifadə eyni olduğuna görə birləşdirilə bilər:

$$
S(\beta) = y^T y - 2 \beta^T X^T y + \beta^T X^T X \beta
$$

---

### 4️⃣ Törəmə götürmək

Minimumu tapmaq üçün $S(\beta)$-ni $\beta$-ya görə törəməsini alırıq:

$$
\frac{\partial S(\beta)}{\partial \beta} = -2 X^T y + 2 X^T X \beta
$$

---

### 5️⃣ Törəməni sıfıra bərabərləşdirmək

Minimum nöqtəni tapmaq üçün:

$$
-2 X^T y + 2 X^T X \beta = 0
$$

---

### 6️⃣ $\beta$-nı həll edirik

Sadələşdiririk:

$$
X^T X \beta = X^T y
$$

---

### 7️⃣ İndi isə $(X^T X)^{-1}$ ilə vururuq

Əgər $X^T X$ tərs matrisə malikdirsə:

$$
\beta = (X^T X)^{-1} X^T y
$$

---

✅ Və bu, **OLS-in bağlanmış (closed-form) həlli** olur.




## $ y = X \beta + \varepsilon $ və  $S(\beta) = \| y - X \beta \|^2$ niyə belə yazılır




### 1️⃣ Sadə xətti reqressiya (1 dəyişənli hal)

Ən sadə modeldə belə yazırıq:

$$
y = \beta_0 + \beta_1 x + \varepsilon
$$

Burada:

* $\beta_0$ → intercept (sabit),
* $\beta_1$ → eğim (slope),
* $x$ → tək müstəqil dəyişən.

Ona görə bəzən belə yazılır:

$$
y - \beta_1 x - \beta_0
$$

---

### 2️⃣ Ümumi çoxölçülü (matrix) hal

Əgər **bir neçə dəyişən** (feature) varsa, onda hər şeyi matrisa şəklində yığmaq rahat olur.
Ona görə **intercept** də daxil olmaqla bütün əmsalları **$\beta$ vektoru**na yığırıq.

Məsələn, əgər sadəcə 1 feature varsa:

$$
X = \begin{bmatrix}
1 & x_1 \\
1 & x_2 \\
\vdots & \vdots \\
1 & x_n
\end{bmatrix}, \quad 
\beta = \begin{bmatrix} \beta_0 \\ \beta_1 \end{bmatrix}
$$

və

$$
y = \begin{bmatrix} y_1 \\ y_2 \\ \vdots \\ y_n \end{bmatrix}
$$

---

### 3️⃣ Modelin matris forması

Bunu belə yaza bilərik:

$$
y = X \beta + \varepsilon
$$

Burada:

* $X$ → həm intercept üçün **1-lər sütunu**, həm də bütün feature-ları daşıyır,
* $\beta$ → bütün əmsallar ($\beta_0, \beta_1, ..., \beta_p$).

---

### 4️⃣ Məqsəd funksiyası

Sadəcə residual (səhv) budur:

$$
y - X \beta
$$

Onun kvadratlarının cəmi isə:

$$
S(\beta) = \| y - X \beta \|^2
$$

---

### 🔑 Nəticə

* **Sadə modeldə** (1 dəyişən): $y - \beta_0 - \beta_1 x$ yazılır.
* **Ümumi modeldə** (çox dəyişən): bu ifadəni qısa yazmaq üçün matris forması istifadə olunur → $ y - X \beta$.
