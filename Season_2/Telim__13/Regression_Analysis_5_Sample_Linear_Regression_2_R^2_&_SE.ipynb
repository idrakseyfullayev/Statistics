## Sample Linear Regression R^2 & SE

```python
from PIL import Image

# Fayl yolunun əsas hissəsi
base_path = r'C:\Users\dell\Desktop\Excel\Handex Excel\Statistics\Season_2\Telim_13\Pictures\Distribution_5_'

# Şəkilləri saxlayacağımız siyahı
image_list = []

# 1-dən 5-ə qədər olan şəkilləri dövr ilə açırıq
for i in range(1, 16):  # 1-dən 5-ə qədər
    full_path = f'{base_path}{i}.jpg'  # Fayl yolunu dinamik olaraq yaradırıq
    img = Image.open(full_path)
    image_list.append(img)  # Şəkili siyahıya əlavə edirik

# Nəticə: images siyahısında bütün yüklənmiş şəkillər olacaq

import matplotlib.pyplot as plt

# Şəkilləri göstərin
for i, image in enumerate(image_list):
    if i == 13:
        plt.figure(figsize=(9, 5))  # Şəkilin ölçüsü
        plt.title("Sample Linear Regression R^2 & SE", size=20)
        plt.imshow(image)
        plt.axis('off')  # Oxları gizlətmək üçün
        plt.show()
        
    else:
        plt.figure(figsize=(25, 5))  # Şəkilin ölçüsü
        plt.title("Sample Linear Regression R^2 & SE", size=20)
        plt.imshow(image)
        plt.axis('off')  # Oxları gizlətmək üçün
        plt.show()
```
