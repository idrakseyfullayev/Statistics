## Poisson Distributions (PMF)

```python
from scipy.stats import poisson

# Parametr
lambda_ = 3  # Orta gecikmə sayı

# P(X = k) ehtimalı
for k in range(0, 21):  # 0-dan 20-yə qədər olan ehtimalları hesablayaq
    prob = poisson.pmf(k, mu=lambda_)
    print(f"P(X = {k}) = {prob*100:.2f}%")
```

## Poisson Distributions (PMF) Visualization

```python
import matplotlib.pyplot as plt
from scipy.stats import poisson

# Parametr
lambda_ = 3  # Orta hadisə sayı

# X və ehtimal dəyərləri
x = range(0, 21)
pmf_values = poisson.pmf(x, mu=lambda_)

# Vizualizasiya
# plt.figure(figsize=(10, 6))
plt.bar(x, pmf_values, color='skyblue', edgecolor='black')
plt.title(f'Poisson Paylanması (λ = {lambda_})')
plt.xlabel('Hadisə sayı (k)')
plt.ylabel('Ehtimal P(X = k)')
plt.grid(True, axis='y', linestyle='--', alpha=0.7)
plt.xticks(x)
plt.show()
```

## Poisson Distributions (CDF)

```python
from scipy.stats import poisson

# Parametr
lambda_ = 3  # Orta gecikmə sayı

# P(X ≤ k) ehtimalı
for k in range(0, 21):  # 0-dan 20-yə qədər olan ehtimalları hesablayaq
    cumulative_prob = poisson.cdf(k, mu=lambda_)
    print(f"P(X ≤ {k}) = {cumulative_prob*100:.2f}%")
```

## Poisson Distributions (CDF) Visualization

```python
import matplotlib.pyplot as plt
from scipy.stats import poisson

# Parametr
lambda_ = 3  # Orta gecikmə sayı

# 0-dan 20-yə qədər olan k üçün kümülatif ehtimalları hesabla
k_values = range(0, 21)
cdf_values = [poisson.cdf(k, mu=lambda_) for k in k_values]

# Bar qrafik çəkmək
# plt.figure(figsize=(12, 6))
plt.bar(k_values, cdf_values, color='skyblue')
plt.xlabel('k (Hadisə sayı)')
plt.ylabel('P(X ≤ k) (Kümülatif ehtimal)')
plt.title('Poisson paylanması üçün Kümülatif Paylanma Funksiyası (CDF)')
plt.ylim(0, 1.05)
plt.grid(axis='y')
plt.show()
```

## Normal Distribution

```python
import numpy as np
from scipy.stats import norm

np.random.seed(1)

# 100 ədəd 0-1 arası təsadüfi ədəd
u = np.random.rand(100)

# Bu təsadüfi ədədləri norm.ppf ilə standart normal paylanmaya çevir
samples = norm.ppf(u)

print(samples)
```

## Normal Distribution Visualization

```python
import numpy as np
from scipy.stats import norm
import matplotlib.pyplot as plt

np.random.seed(1)

# 100 ədəd 0-1 arası təsadüfi ədəd
u = np.random.rand(100)

# norm.ppf ilə standart normal paylanmaya çevir
samples = norm.ppf(u)

# Histogram çəkmək
# plt.figure(figsize=(10, 6))
plt.hist(samples, bins=20, color='purple', edgecolor='black')
plt.xlabel('Dəyər')
plt.ylabel('Tezlik')
plt.title('Standart Normal Paylanmadan 100 Nümunənin Histogramı')
plt.grid(axis='y')
plt.show()
```

## Excel-də paylanma və tərs funksiyalar: NORM, BINOM, POISSON


### ✅ `NORM.DIST` → **ehtimal qaytarır**

`
=NORM.DIST(x, mean, std_dev, TRUE)
`

🔹 Verilən **x dəyərinə** qədər olan **ehtimalı** (P(X ≤ x)) qaytarır.

---

### ✅ `NORM.INV` → **dəyər qaytarır**

`
=NORM.INV(probability, mean, std_dev)
`

🔹 Verilən **ehtimala** uyğun olan **x dəyərini** qaytarır.

---

### 🔁 Sadə cədvəl:

| Funksiya    | Giriş       | Çıxış        | Mənası              |
| ----------- | ----------- | ------------ | ------------------- |
| `NORM.DIST` | **x**       | **ehtimal**  | P(X ≤ x)            |
| `NORM.INV`  | **ehtimal** | **x** dəyəri | Bu ehtimala uyğun x |

---

### ✅ `BINOM.DIST` → **ehtimal qaytarır**

`
=BINOM.DIST(x, trials, prob, TRUE)
`

🔹 Verilən `x` uğur sayı üçün **P(X ≤ x)** ehtimalını qaytarır (CDF).
TRUE → kümülatif ehtimal, FALSE → P(X = x) (sırf həmin ehtimal).

---

### ✅ `BINOM.INV` → **dəyər qaytarır**

`
=BINOM.INV(trials, prob, probability)
`

🔹 Verilən **ehtimala** görə **minimum x** dəyərini qaytarır ki, P(X ≤ x) ≥ ehtimal olsun.

---

### 🔁 Sadə cədvəl:

| Funksiya     | Giriş       | Çıxış        | Mənası                       |
| ------------ | ----------- | ------------ | ---------------------------- |
| `BINOM.DIST` | **x**       | **ehtimal**  | P(X ≤ x) və ya P(X = x)      |
| `BINOM.INV`  | **ehtimal** | **x** dəyəri | Bu ehtimala uyğun ən kiçik x |

---

### ✅ `POISSON.DIST` → **ehtimal qaytarır**

`
=POISSON.DIST(x, mean, cumulative)
`

🔹 Verilmiş `x` üçün **P(X ≤ x)** (TRUE) və ya **P(X = x)** (FALSE) ehtimalını qaytarır.
Bu funksiya **Poisson paylanması** üçün istifadə olunur (məsələn, 1 saatda gələn zəng sayı).

---

> **Qeyd:** Excel-də Poisson üçün **invers funksiya yoxdur** (`POISSON.INV` yoxdur).
> Əgər ehtimala uyğun `x` dəyəri tapmaq istəsən, əl ilə təkrarlama (iterasiya) etməlisən və ya VBA/Python istifadə edə bilərsən.

---

### 🔁 Sadə cədvəl:

| Funksiya       | Giriş    | Çıxış       | Mənası                  |
| -------------- | -------- | ----------- | ----------------------- |
| `POISSON.DIST` | **x**    | **ehtimal** | P(X ≤ x) və ya P(X = x) |
| `POISSON.INV`  | ❌ yoxdur | ❌ yoxdur    | Excel-də mövcud deyil   |
