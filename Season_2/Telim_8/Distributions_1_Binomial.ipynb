## Binom Distributions (PMF)

```python
from scipy.stats import binom

# Parametrlər
n = 22      # sınaq sayı
p = 4/22     # uğur ehtimalı
# k uğurun sayı (məsələn: 3 Gecikmə)

# P(X = k) ehtimalı
for k in range(n + 1):
    prob = binom.pmf(k, n, p)
    print(f"P(X = {k}) = {prob*100:.1f}%")
```

## Binom Distributions(PMF) Visualation

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter
import numpy as np

x = range(n+1)
pmf_values = binom.pmf(x, n, p)

plt.bar(x, pmf_values, color='skyblue')
plt.xlabel('Uğur sayı (k)')
plt.ylabel('P(X = k)')
plt.title(f'Binom Paylanması (n={n}, p={p:.3f})')
plt.grid(True, axis='y', linestyle='--', alpha=0.7)

plt.xticks(x)

# Y oxu: hər 5% bir göstər
yticks = np.arange(0, 0.30, 0.05)  # 0.00, 0.05, 0.10, ..., 0.25
plt.yticks(yticks)
plt.gca().yaxis.set_major_formatter(PercentFormatter(1.0))  # 1.0 → 100% deməkdir
plt.ylim(0, 0.30)  # Y oxunu 0%-30% aralığında göstər

plt.show()
```

## Binom Distributions (CDF)

```python
from scipy.stats import binom

# Parametrlər
n = 22      # sınaq sayı
p = 4/22     # uğur ehtimalı
# k uğurun sayı (məsələn: 3 Gecikmə)

# P(X = k) ehtimalı
for k in range(n + 1):
    prob = binom.cdf(k, n, p)
    print(f"P(X ≤ {k}) = {prob*100:.1f}%")
```

## Binom Distributions (CDF) Visaullation

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import PercentFormatter
from scipy.stats import binom
import numpy as np

n = 22
p = 4 / 22

x = range(n + 1)
cdf_values = binom.cdf(x, n, p)  # BÜTÜN k-lər üçün CDF-ləri hesabla

plt.bar(x, cdf_values, color='skyblue')
plt.xlabel('Uğur sayı (k)')
plt.ylabel('P(X ≤ k)')
plt.title(f'Binom Paylanması - Kümülatif (n={n}, p={p:.3f})')

plt.grid(True, axis='y', linestyle='--', alpha=0.7)
plt.xticks(x)

# Y oxu: hər 5% bir göstər
yticks = np.arange(0, 1.05, 0.05)  # 0.00-dan 1.00-a qədər
plt.yticks(yticks)
plt.gca().yaxis.set_major_formatter(PercentFormatter(1.0))
plt.ylim(0, 1.0)  # CDF 100%-ə qədər gedir

plt.show()
```

## Bernoulli sınağında Dispersiya



### 🔹 **1. Bernoulli sınağı nədir?**

Bernoulli sınağı — yalnız **iki mümkün nəticəsi olan** bir təcrübədir:

* Məsələn: Yazı → 1, Tura → 0
* **Uğur (1)** ehtimalı $p$,
* **Uğursuzluq (0)** ehtimalı $1 - p$ olur.

---

### 🔹 **2. Təsadüfi dəyişən və gözlənən qiymət**

Bu sınaq üçün təsadüfi dəyişən $X$ belə təyin olunur:

$$
X = 
\begin{cases}
1 & \text{ehtimal } p \\
0 & \text{ehtimal } 1 - p
\end{cases}
$$

📌 **Riyazi gözləmə (Expected Value)**:

$$
E(X) = 1 \cdot p + 0 \cdot (1 - p) = p
$$

---

### 🔹 **3. Dispersiya (Varians) nədir?**

Dispersiya belə hesablanır:

$$
\text{Var}(X) = E[X^2] - (E[X])^2
$$

Bernoulli sınağında:

$$
X^2 = X \quad \text{(çünki 0² = 0, 1² = 1)}
$$

Ona görə:

$$
E[X^2] = E[X] = p
$$

$$
\Rightarrow \text{Var}(X) = p - p^2 = p(1 - p)
$$

---

### 🔹 **Nəticə:**

Bernoulli sınağında dispersiya:

$$
\boxed{\text{Var}(X) = p(1 - p)}
$$

Bu ifadə, təsadüfi dəyişənin nə qədər dəyişkən olduğunu göstərir. Əgər $p$ çox kiçik və ya çox böyükdürsə, dispersiya azalır. Ən böyük dispersiya isə $p = 0.5$ olanda olur (çünki $0.5 \cdot 0.5 = 0.25$).


## Bernoulli sınaqlarının tətbiqi

Ailədə 3 uşaq var və hər bir uşaq **qız** və ya **oğlan** ola bilər. Tutaq ki, **bir uşağın qız olma ehtimalı** $p$-dir (məsələn, adətən 0.5 götürülür). Bu vəziyyətdə məsələ **Bernoulli sınaqlarının** (daha dəqiqi **Binom paylanmasının**) tətbiqidir.

---

### 🔹 Tapşırıq:

3 uşağın **0, 1, 2 və 3** qız olması ehtimallarını **Bernoulli düsturu ilə** hesabla.

---

### 🔸 1. **Düstur: Binom paylanması**

Ailədə 3 uşaq → 3 sınaq

Binomial ehtimal düsturu:

$$
P(X = k) = \binom{n}{k} \cdot p^k \cdot (1 - p)^{n - k}
$$

Burada:

* $n = 3$ (uşaq sayı),
* $k = 0, 1, 2, 3$ (qız sayı),
* $p$ = bir uşağın qız olma ehtimalı (məsələn, 0.5).

---

### 🔸 2. **Ehtimalların hesablanması** (əgər $p = 0.5$ olsa):

### ✅ $P(0 \text{ qız})$

$$
P(X = 0) = \binom{3}{0} \cdot 0.5^0 \cdot 0.5^3 = 1 \cdot 1 \cdot 0.125 = \boxed{0.125}
$$

### ✅ $P(1 \text{ qız})$

$$
P(X = 1) = \binom{3}{1} \cdot 0.5^1 \cdot 0.5^2 = 3 \cdot 0.5 \cdot 0.25 = \boxed{0.375}
$$

### ✅ $P(2 \text{ qız})$

$$
P(X = 2) = \binom{3}{2} \cdot 0.5^2 \cdot 0.5^1 = 3 \cdot 0.25 \cdot 0.5 = \boxed{0.375}
$$

### ✅ $P(3 \text{ qız})$

$$
P(X = 3) = \binom{3}{3} \cdot 0.5^3 \cdot 0.5^0 = 1 \cdot 0.125 \cdot 1 = \boxed{0.125}
$$

---

### 🔸 3. **Nəticə cədvəli:**

| Qız sayı $$k$$ | Ehtimal $$P(X = k)$$ |
| ------------ | ------------------ |
| 0            | 0.125              |
| 1            | 0.375              |
| 2            | 0.375              |
| 3            | 0.125              |

Bunların cəmi də:

$$
0.125 + 0.375 + 0.375 + 0.125 = 1
$$




## Bernoulli və Binom Paylanmasının Fərqi



### 🔷 1. **Bernoulli Paylanması**

### ✔️ **Düstur**:

$$
P(X = x) = p^x \cdot (1 - p)^{1 - x}, \quad x \in \{0, 1\}
$$

Burada:

* $X$ – bir uşağın qız olması (1: qız, 0: oğlan),
* $p$ – qız olma ehtimalı.

---

### ✔️ **Misal**:

**Ailədə 1 uşaq var**, bu uşağın **qız olma ehtimalı** $p = 0.5$-dir.

* $P(X = 1)$ — qız olma ehtimalı:

$$
P(X = 1) = (0.5)^1 \cdot (1 - 0.5)^{0} = 0.5 \cdot 1 = \boxed{0.5}
$$

* $P(X = 0)$ — oğlan olma ehtimalı:

$$
P(X = 0) = (0.5)^0 \cdot (1 - 0.5)^1 = 1 \cdot 0.5 = \boxed{0.5}
$$

---

### 🔷 2. **Binom Paylanması**

### ✔️ **Düstur**:

$$
P(X = k) = \binom{n}{k} \cdot p^k \cdot (1 - p)^{n - k}
$$

Burada:

* $n$ — uşaq (sınaq) sayı,
* $k$ — qızların (uğurların) sayı,
* $p$ — bir uşağın qız olma ehtimalı.

---

### ✔️ **Misal**:

**Ailədə 3 uşaq var**, qız olma ehtimalı $p = 0.5$-dir.
Sual: **3 uşaqdan 2-sinin qız olma ehtimalı** nədir?

Burada:

* $n = 3$,
* $k = 2$,
* $p = 0.5$

$$
P(X = 2) = \binom{3}{2} \cdot (0.5)^2 \cdot (0.5)^1 = 3 \cdot 0.25 \cdot 0.5 = \boxed{0.375}
$$

---

### 🔔 **Nəticə fərqi:**

| Paylanma növü | Sınaq sayı | Düstur                                    | Misal                                |
| ------------- | ---------- | ----------------------------------------- | ------------------------------------ |
| **Bernoulli** | **1 sınaq**    | $$P(X = x) = p^x (1-p)^{1-x}$$              | 1 uşaq → qız olma ehtimalı $$P = 0.5$$ |
| **Binomial**  | **n sınaq**  | $$P(X = k) = \binom{n}{k} p^k (1-p)^{n-k}$$ | 3 uşaqdan 2 qız → $$P = 0.375$$        |



## Paylanma funksiyası və Sıxlıq funksiyası (PMF, PDF, CDF)

* **Paylanma funksiyası (F(x))** = ehtimalların **toplanmış** (kümülatif) halıdır.
* **Sıxlıq funksiyası (f(x))** isə **birbaşa ehtimal (diskret)** və ya **sıxlıq (kəsilməz)** verir — **toplanmır**, öz-özlüyündə ehtimal deyil (kəsilməzdə), yalnız **inteqralı ehtimal verir.**

---

## 🔍 Geniş izah:

### 🟢 **Diskret hal (məsələn: X = 2, 3, 5)**

#### 1. **Sıxlıq funksiyası (PMF — f(x))**

* Hər bir **dəqiq x** üçün ehtimal verir:

  $$
  f(2) = P(X=2) = 0.2, \quad f(3) = 0.5, \quad f(5) = 0.3
  $$

* **Toplamırıq!** f(x) özü bir ehtimaldır. Amma **bütün f(x)-lərin cəmi 1 olur**:

  $$
  \sum f(x) = 0.2 + 0.5 + 0.3 = 1
  $$

#### 2. **Paylanma funksiyası (CDF — F(x))**

* CDF **toplanmış ehtimallardır**:

  $$
  F(x) = P(X \le x) = \sum_{t \le x} f(t)
  $$

  Məsələn:

  * \$F(2) = P(X \le 2) = 0.2\$
  * \$F(3) = P(X \le 3) = 0.2 + 0.5 = 0.7\$
  * \$F(5) = 0.2 + 0.5 + 0.3 = 1\$

### 🔁 **Yəni:**

* **f(x)** → ayrılıqda ehtimal
* **F(x)** → həmin ehtimalların yığılmış (toplanmış) versiyası

---

### 🔵 **Kəsilməz hal (məsələn: X \~ N(μ=0, σ=1))**

#### 1. **Sıxlıq funksiyası (PDF — f(x))**

* f(x) burada **bir nöqtə üçün ehtimal deyil**, sadəcə **sıxlığı** göstərir.
* Ehtimalı hesablamaq üçün **interval üzrə inteqral** aparılır:

  $$
  P(a \le X \le b) = \int_a^b f(x)\, dx
  $$

🔹 **Nümunə:**
X \~ N(0, 1) — standart normal paylanma.
Onda:

* f(0) = 0.3989 (maksimum sıxlıq), amma \$P(X = 0) = 0\$ çünki **nöqtə ehtimalı** kəsilməz paylanmada **0-dır**.
* \$P(-1 \le X \le 1) = \int\_{-1}^{1} f(x) dx \approx 0.6826\$

#### 2. **Paylanma funksiyası (CDF — F(x))**

* Burada da F(x) ehtimalların "yığılmış" formasıdır:

  $$
  F(x) = P(X \le x) = \int_{-\infty}^{x} f(t)\, dt
  $$

🔹 **Nümunə:**

* \$F(0) = 0.5\$ (standart normalda simmetrikdir)
* \$F(1) \approx 0.8413\$
* \$F(-1) \approx 0.1587\$

---

## 🔚 Nəticə:

| Funksiya            | Diskret (PMF)             | Kəsilməz (PDF)                         |
| ------------------- | ------------------------- | -------------------------------------- |
| **f(x)** (sıxlıq)   | Ehtimaldır, **toplanmaz** | Sıxlıqdır, **inteqralı ehtimal verir** |
| **F(x)** (paylanma) | f(x)-lər **toplanır**     | f(x)-in **inteqralı alınır**           |

---

## ✅ Sənin sualına bir cümləlik cavab:

> Paylanma funksiyasında ehtimallar **toplanır**, çünki bu funksiyanın məqsədi "toplanmış ehtimalı" — yəni \$P(X \le x)\$-i verməkdir.
> Sıxlıq funksiyası isə **nəticənin öz ehtimalını/sıxlığını** göstərir, yığılma yoxdur.
