## Confidence interval
- **Ortalaması 50 olan, standart sapması 10 və nümunə sayı 25 olan bir məlumat üçün 95% güvən intervalını verəcək.**

```python
import math
from scipy.stats import norm

# Parametrlər
mean = 50
std_dev = 10
n = 25
alpha = 0.05

# Kritiki z dəyəri (1 - α/2 üçün)
z = norm.ppf(1 - alpha/2)

# Margin of error (səhv payı)
margin = z * (std_dev / math.sqrt(n))

# Etibarlılıq intervalı
lower = mean - margin
upper = mean + margin

print(f"Etibarlılıq intervalı: [{lower:.2f}, {upper:.2f}]")
```
