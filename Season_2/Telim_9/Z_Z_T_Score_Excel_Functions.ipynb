## 📊 Excel-də Z ilə bağlı funksiyalar – CƏDVƏL



### ✅ **Z-test və normal paylanma funksiyaları**

| ✅ **Funksiya**                       | 📌 **Tətbiq sahəsi**                              | 🔍 **Nə edir?**                                    |
| ------------------------------------ | ------------------------------------------------- | -------------------------------------------------- |
| `Z.TEST(array, x)`                   | Z-test (bir nümunə, orta dəyərlə müqayisə)        | **p-value** qaytarır (tək tərəfli)                 |
| `=(x̄ - μ₀) / (σ/√n)`                | Z-score hesablamaq (manual)                       | Z test statistikası (score) qaytarır               |
| `NORM.S.DIST(z, TRUE)`               | Z-score üçün p-value                              | Z-score-a uyğun **CDF (kumulyativ ehtimal)** verir |
| `NORM.S.DIST(z, FALSE)`              | Z-score üçün sıxlıq                               | Z-score-a uyğun **PDF (sıxlıq)** verir             |
| `NORM.DIST(x, mean, std_dev, TRUE)`  | Normal paylanma ehtimalı (CDF)                    | Verilən **x** üçün **kümülatif ehtimal** verir     |
| `NORM.DIST(x, mean, std_dev, FALSE)` | Normal paylanma sıxlığı (PDF)                     | Verilən **x** üçün **sıxlıq funksiyası** verir     |
| `NORM.INV(p, mean, std_dev)`         | Normal paylanmada x dəyəri                        | Verilən ehtimala uyğun **x dəyərini** qaytarır     |
| `NORM.S.INV(p)`                      | Z-score tapmaq (əvvəlcədən verilmiş ehtimal üçün) | Verilən ehtimala uyğun **z-score** qaytarır        |



### ✅ **📌 Nümunə sətirlər**

| Test             | Formul                                                         | Açıklama                             |
| ---------------- | -------------------------------------------------------------- | ------------------------------------ |
| Z-test (p-value) | `=Z.TEST(A1:A30, 100)`                                         | Nümunə 100 ortalama ilə fərqlidirmi? |
| Norm. ehtimal    | `=NORM.DIST(105, 100, 15, TRUE)`                               | 105 və ya daha aşağı olma ehtimalı   |
| Norm. sıxlıq     | `=NORM.DIST(105, 100, 15, FALSE)`                              | 105 üçün sıxlıq (PDF)                |
| Z-score tapmaq   | `=NORM.S.INV(0.975)`                                           | 97.5% ehtimala uyğun z-score         |



## Python-da Z ilə bağlı funksiyalar – CƏDVƏL



| ✅ **Excel Funksiyası**               | 📌 **Tətbiq sahəsi**                              | 🔍 **Nə edir?**                                    | 🐍 **Python funksiyası (kitabxana)**               | 🐍 **İstifadə nümunəsi**                                 |
| ------------------------------------ | ------------------------------------------------- | -------------------------------------------------- | -------------------------------------------------- | -------------------------------------------------------- |
| `Z.TEST(array, x)`                   | Z-test (bir nümunə, orta dəyərlə müqayisə)        | **p-value** qaytarır (tək tərəfli)                 | `scipy.stats.norm.sf(z)` + z hesablaması           | `z = (x̄ - μ₀) / (σ / np.sqrt(n))` <br> `p = norm.sf(z)` |
| `=(x̄ - μ₀) / (σ/√n)`                | Z-score hesablamaq (manual)                       | Z test statistikası (score) qaytarır               | `manual hesablamadır`                              | `z = (x̄ - μ₀) / (σ / np.sqrt(n))`                       |
| `NORM.S.DIST(z, TRUE)`               | Z-score üçün p-value                              | Z-score-a uyğun **CDF (kumulyativ ehtimal)** verir | `scipy.stats.norm.cdf(z)`                          | `p = norm.cdf(z)`                                        |
| `NORM.S.DIST(z, FALSE)`              | Z-score üçün sıxlıq                               | Z-score-a uyğun **PDF (sıxlıq)** verir             | `scipy.stats.norm.pdf(z)`                          | `density = norm.pdf(z)`                                  |
| `NORM.DIST(x, mean, std_dev, TRUE)`  | Normal paylanma ehtimalı (CDF)                    | Verilən **x** üçün **kümülatif ehtimal** verir     | `scipy.stats.norm.cdf(x, loc=mean, scale=std_dev)` | `p = norm.cdf(x, loc=mean, scale=std_dev)`               |
| `NORM.DIST(x, mean, std_dev, FALSE)` | Normal paylanma sıxlığı (PDF)                     | Verilən **x** üçün **sıxlıq funksiyası** verir     | `scipy.stats.norm.pdf(x, loc=mean, scale=std_dev)` | `density = norm.pdf(x, loc=mean, scale=std_dev)`         |
| `NORM.INV(p, mean, std_dev)`         | Normal paylanmada x dəyəri                        | Verilən ehtimala uyğun **x dəyərini** qaytarır     | `scipy.stats.norm.ppf(p, loc=mean, scale=std_dev)` | `x = norm.ppf(p, loc=mean, scale=std_dev)`               |
| `NORM.S.INV(p)`                      | Z-score tapmaq (əvvəlcədən verilmiş ehtimal üçün) | Verilən ehtimala uyğun **z-score** qaytarır        | `scipy.stats.norm.ppf(p)`                          | `z = norm.ppf(p)`                                        |

---

### Qısa izah:

* `norm.cdf()` — kümülatif ehtimal funksiyası (CDF)
* `norm.pdf()` — ehtimal sıxlıq funksiyası (PDF)
* `norm.ppf()` — inverz kümülatif ehtimal (Quantile function)




## Excel-də Confidence Interval 



### ✅ **Excel-də Confidence Interval və Statistik Funksiyalar**

| ✅ **Funksiya / Düstur**               | 📌 **Tətbiq sahəsi**                              | 🔍 **Nə edir?**                                                       |
| ------------------------------------- | ------------------------------------------------- | --------------------------------------------------------------------- |
| `=CONFIDENCE.NORM(alpha, std_dev, n)` | Z-da əsaslı etibarlılıq intervalı                 | Verilən **etibarlılıq səviyyəsinə** uyğun **marge of error** qaytarır |
| `=CONFIDENCE.T(alpha, std_dev, n)`    | T-da əsaslı etibarlılıq intervalı                 | Kiçik nümunə üçün (n<30) **t-score əsaslı marge of error** verir      |
| `=(x̄ - μ₀) / (σ/√n)`                 | Z-score hesabı                                    | Əl ilə Z statistikası hesablayır                                      |
| `=(x̄ - μ₀) / (s/√n)`                 | T-score hesabı                                    | Əl ilə T statistikası hesablayır                                      |
| `=x̄ ± Z * σ/√n`                      | Güvən intervalı (Z ilə, əl ilə)                   | Etibarlılıq intervalını manual hesablamaq                             |
| `=x̄ ± T * s/√n`                      | Güvən intervalı (T ilə, əl ilə)                   | Kiçik nümunələr üçün manual interval                                  |




## Python-da Confidence Interval



| ✅ **Excel Funksiyası / Düstur**       | 📌 **Tətbiq sahəsi**                            | 🔍 **Nə edir?**                                                     | 🐍 **Python funksiyası / düsturu**                                | 🐍 **İstifadə nümunəsi**                                                                                |
| ------------------------------------- | ----------------------------------------------- | ------------------------------------------------------------------- | ----------------------------------------------------------------- | ------------------------------------------------------------------------------------------------------- |
| `=CONFIDENCE.NORM(alpha, std_dev, n)` | Z əsaslı etibarlılıq intervalı (normal paylama) | Verilən **etibarlılıq səviyyəsi** üçün **margin of error** qaytarır | `scipy.stats.norm.ppf(1 - alpha/2) * (std_dev / np.sqrt(n))`      | `python<br>from scipy.stats import norm<br>margin = norm.ppf(1 - alpha/2) * (std_dev/np.sqrt(n))<br>`   |
| `=CONFIDENCE.T(alpha, std_dev, n)`    | T əsaslı etibarlılıq intervalı (kiçik nümunə)   | Kiçik nümunələr üçün (n<30) **t-score** əsaslı margin of error      | `scipy.stats.t.ppf(1 - alpha/2, df=n-1) * (std_dev / np.sqrt(n))` | `python<br>from scipy.stats import t<br>margin = t.ppf(1 - alpha/2, df=n-1) * (std_dev/np.sqrt(n))<br>` |
| `=(x̄ - μ₀) / (σ/√n)`                 | Z-score hesabı                                  | Z test statistikası (manual hesab)                                  | `z = (x_bar - mu_0) / (sigma / np.sqrt(n))`                       | `python<br>z = (x_bar - mu_0) / (sigma / np.sqrt(n))<br>`                                               |
| `=(x̄ - μ₀) / (s/√n)`                 | T-score hesabı                                  | T test statistikası (manual hesab)                                  | `t = (x_bar - mu_0) / (s / np.sqrt(n))`                           | `python<br>t = (x_bar - mu_0) / (s / np.sqrt(n))<br>`                                                   |
| `=x̄ ± Z * σ/√n`                      | Güvən intervalı (Z ilə, manual)                 | Etibarlılıq intervalını Z-score ilə hesablamaq                      | `(x_bar - margin, x_bar + margin)`                                | `python<br>lower = x_bar - margin<br>upper = x_bar + margin<br>`                                        |
| `=x̄ ± T * s/√n`                      | Güvən intervalı (T ilə, manual)                 | Kiçik nümunələr üçün T-score ilə interval                           | `(x_bar - margin, x_bar + margin)`                                | `python<br>lower = x_bar - margin<br>upper = x_bar + margin<br>`                                        |

---

### Qeyd:

* `alpha` — əhəmiyyət səviyyəsi (məsələn, 0.05 üçün 95% confidence)
* `std_dev` — populyasiya standart sapması (Z əsaslı hesablama üçün)
* `s` — nümunə standart sapması (T əsaslı hesablama üçün)
* `n` — nümunə sayı
* `df = n - 1` — sərbəstlik dərəcəsi T paylanması üçün

## 📊 Excel-də T ilə bağlı funksiyalar – CƏDVƏL


### ✅ **T-test və t-paylanma funksiyaları**

| ✅ **Funksiya**                        | 📌 **Tətbiq sahəsi**               | 🔍 **Nə edir?**                                                          |
| ------------------------------------- | ---------------------------------- | ------------------------------------------------------------------------ |
| `T.TEST(array1, array2, tails, type)` | İki nümunə üçün t-test             | İki qrupun ortaları arasında fərq olub-olmadığını yoxlayır (**p-value**) |
