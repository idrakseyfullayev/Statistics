```python
import numpy as np
import matplotlib.pyplot as plt

# Random ədədlər
np.random.seed(0)  # eyni nəticə üçün
a = np.random.randint(1, 6, 500)  # 1-5 arası
b = np.random.randint(1, 6, 500)

# Setir-setir ortalamalar
pair_means = (a + b) / 2

# Vizualizasiya - Histogram
plt.figure(figsize=(10, 5))
plt.hist(pair_means, bins=np.arange(1, 7, 0.5), edgecolor='black', color='skyblue')
plt.title("Setir-Setir Ortalamaların Histogramı")
plt.xlabel("Orta Dəyər")
plt.ylabel("Sıklıq")
plt.xticks(np.arange(1, 5.5, 0.5))
plt.grid(True, axis='y')
plt.show()
```

## Bias-Variance Tradeoff

**Bias-Variance Tradeoff** (Bəəz-Varyans Tarazlığı) — **machine learning** və **statistik modelləşdirmədə** modellərin **dəqiqliyi** və **ümumiləşdirmə qabiliyyəti** arasında balansı izah edən **fundamental anlayışdır**.

---

### ✅ Sadə dillə tərifi:

Bir modelin **həddindən artıq sadə** olması (az öyrənməsi) və ya **həddindən artıq mürəkkəb** olması (çox öyrənməsi) nəticəsində **səhv** (error) yaranır. Bu səhvlər iki əsas səbəbdən yaranır:

### 🔹 1. **Bias (Sistematik qərəz)**

* Model **həddindən artıq sadədir** (məsələn, linear model).
* Həqiqi əlaqəni **qaçırır**.
* **Underfitting** yaradır.
* Nəticə: **Yüksək səhv həm təlimdə, həm də testdə.**

### 🔹 2. **Variance (Dəyişkənlik)**

* Model **çox mürəkkəbdir** (məsələn, çox dərin decision tree).
* Təlim datasını **çox yaxşı əzbərləyir**.
* Test datasında **zəif performans** göstərir.
* **Overfitting** olur.

---

### 🔁 **Bias-Variance Tradeoff nədir?**

Model qurarkən məqsəd **bias və variance arasında optimal balans** tapmaqdır.

* **Sadə model** → Yüksək bias, Aşağı variance
* **Çox mürəkkəb model** → Aşağı bias, Yüksək variance
* **Ən yaxşı model** → **Orta bias və orta variance** → **Ən az ümumi səhv**

---

### 📊 Vizual təsvir (mənə istəsən qrafiklə də göstərə bilərəm):

| Model növü         | Bias   | Variance | Ümumi Error |
| ------------------ | ------ | -------- | ----------- |
| Sadə (Underfit)    | Yüksək | Aşağı    | Yüksək      |
| Optimal            | Orta   | Orta     | **Aşağı**   |
| Mürəkkəb (Overfit) | Aşağı  | Yüksək   | Yüksək      |

---

### 📌 Nəticə:

**Bias-Variance Tradeoff**, **ən az səhv ilə ümumiləşdirə bilən** model tapmaq üçündür.
