## Sample Size

> Excel-də **verilən sigma (σ = 45)**, **etibarlılıq səviyyəsi (məsələn 95%)** və **marge of error (xəta payı, məsələn ±5)** ilə **lazımi nümunə sayını (sample size)** hesablamaq.


### ✅ **Nümunə Sayı (Sample Size) Düsturu**

Əgər **populyasiya standart sapması (σ)** məlumdursa, aşağıdakı düsturdan istifadə olunur:

$$
n = \left(\frac{Z \cdot \sigma}{E}\right)^2
$$

Burada:

* **Z** — Z-score (95% üçün ≈ 1.96)
* **σ** — Standart sapma (burada 45)
* **E** — Marge of error (məsələn 5)

---

### ✅ **Excel-də Formula (manual)**

Tutaq ki:

* `sigma = 45`
* `error = 5`
* `Z = NORM.S.INV(1 - 0.05/2)` → bu da 1.96 verir

**Formula:**

`
=(NORM.S.INV(1 - 0.05/2) * 45 / 5)^2
`

---

### 📌 **Nümunə:**

`
=(1.96 * 45 / 5)^2
`

**Nəticə:** `n ≈ 311.17` → yəni **ən azı 312 nəfərlik nümunə** lazımdır.
